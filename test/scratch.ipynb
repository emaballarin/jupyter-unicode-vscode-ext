## 5. Ctrl+Space

`<Ctrl+Space>` is VS Code's own trigger for the suggestion list, and this extension
is an ordinary provider, so it answers there too. What differs is when each fires.

In the cell below, check:

- `\bet` then `<Ctrl+Space>` offers `\beta`, exactly as `<Tab>` does.
- Select a couple of lines so the selection ends just after a `\name`. `<Tab>`
  indents the block; `<Ctrl+Space>` still opens the list.
- After `C:\Users`, `<Tab>` indents and `<Ctrl+Space>` offers nothing from us.
- In a `.txt` file — where nothing else provides completions — `<Ctrl+Space>`
  still works, because this extension is what makes VS Code consider the binding
  live there at all.
